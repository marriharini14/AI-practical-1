```python
#Tressure Hunt
print("Hello, Welcome to Tressure Hunt")
print("Your mission is to find Tressure")

MAP = {
    "A": ["B", "C"],
    "B": ["D"],
    "C": ["D"],
    "D": ["G"],
    "G": [],
}

print("Our Map: ")
print("                  A (Start)")
print("                /  \\ ")
print("               B     C ")
print("               \\   / ")
print("                  D ")
print("                  |" )
print("                  G (Tressure )" )
print("Doors for each rooms:", MAP)
```

```python
from typing import OrderedDict
def bfs(Start,goal):
  to_do =[Start]
  visited = []
  order = []

  while to_do:
    room = to_do.pop()

    if room in visited:
      continue
    visited.append(room)
    order.append(room)

    if room == goal:
      return order

    for next_room in MAP[room]:
      to_do.append(next_room)

    return order
order= bfs("A" ,"G")
print("BFS checked the rooms in this order:",order)
print("Numbers of rooms BFS checked:",len(order))
```

```python
def DFS(start,goal):
    to_do = [start]
    visited = []
    order = []

    while to_do:
      room = to_do.pop()

      if room in visited:
         continue
      visited.append(room)
      order.append(room)

      if room == goal:
          return order
      for nxt in MAP[room]:
          to_do.append(nxt)
    return order



order= DFS("A" ,"G")
print("DFS checked the rooms in this order:",order)
print("Numbers of rooms DFS checked:",len(order))
```
